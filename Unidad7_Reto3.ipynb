```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 3
reciclaje = 80 #@param {type:"slider", min:0, max:100, step:10}
energia = 70 #@param {type:"slider", min:0, max:100, step:10}
seguridad = 90 #@param {type:"slider", min:0, max:100, step:10}

score = (reciclaje + energia + seguridad) / 3
print(f"Score Global ODS: {score:.1f}/100 -> {'APROBADO' if score>=75 else 'RECHAZADO'}")
plt.bar(["Score ODS"], [score], color="#00A859" if score>=75 else "#E63946"); plt.axhline(75, color="black", linestyle="--"); plt.ylim(0,100); plt.show()
```
