```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 3
marcas_asociadas = 5 #@param {type:"slider", min:2, max:15, step:1}
capacidad_marca = 200 #@param {type:"number"}

total_unidades = marcas_asociadas * capacidad_marca
ocupacion_pct = min(100, (total_unidades / 1200) * 100)

print(f"Producción Consolidada: {total_unidades} unidades | Ocupación Contenedor: {ocupacion_pct:.1f}%")
plt.bar(["Ocupación Contenedor (%)"], [ocupacion_pct], color="#00A859" if ocupacion_pct>=80 else "#E63946"); plt.ylim(0, 100); plt.show()
```
