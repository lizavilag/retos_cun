```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 2
quejas_totales = 500 #@param {type:"number"}
inversion_compensacion_por_cliente = 50 #@param {type:"slider", min:0, max:200, step:10}

retencion_est = min(100, (inversion_compensacion_por_cliente / 150) * 100)
costo_total = quejas_totales * inversion_compensacion_por_cliente

print(f"Costo Total Crisis: ${costo_total} | Estimación Retención: {retencion_est:.1f}%")
plt.pie([retencion_est, 100-retencion_est], labels=["Clientes Retenidos", "Clientes Perdidos"], colors=["#00A859","#E63946"], autopct="%1.1f%%")
plt.show()
```
