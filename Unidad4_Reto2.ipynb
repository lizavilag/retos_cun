```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 2
desperdicio_ton = 20 #@param {type:"number"}
eficiencia_pct = 45 #@param {type:"slider", min:10, max:90, step:5}

ton_salvadas = desperdicio_ton * (eficiencia_pct/100)
ahorro_mes = ton_salvadas * 350

print(f"Residuos Evitados: {ton_salvadas:.1f} Ton/mes | Ahorro: ${ahorro_mes:.2f}/mes")
plt.bar(["Desperdicio Original", "Nuevo Desperdicio"], [desperdicio_ton, desperdicio_ton - ton_salvadas], color=["#E63946", "#00A859"]); plt.show()
```
