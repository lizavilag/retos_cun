```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 2
horas_reuniones = 18 #@param {type:"slider", min:2, max:30, step:2}

fatiga = min(100, horas_reuniones * 4.2)
foco = max(0, 40 - horas_reuniones - 5)
print(f"Fatiga Digital: {fatiga:.0f}/100 | Horas Foco: {foco} hrs/sem")
plt.bar(["Horas Foco", "Horas Reunión"], [foco, horas_reuniones], color=["#00A859","#E63946"]); plt.show()
```
