```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 3
recirculacion_pct = 60 #@param {type:"slider", min:10, max:95, step:5}
base_m3 = 1000 #@param {type:"number"}

nuevo_m3 = base_m3 * (1 - recirculacion_pct/100)
ahorro = (base_m3 - nuevo_m3) * 4.5
print(f"Nuevo Consumo: {nuevo_m3:.0f} m3 | Ahorro Mensual: ${ahorro:.2f}")
plt.bar(["Consumo Previo", "Nuevo Consumo"], [base_m3, nuevo_m3], color=["#E63946","#00A859"]); plt.show()
```
