```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 2
dias_stock = 10 #@param {type:"slider", min:2, max:30, step:2}

costo_stock = dias_stock * 120
riesgo_paro = 15000 / dias_stock
print(f"Costo Almacenamiento: ${costo_stock} | Riesgo Paro: ${riesgo_paro:.2f}")
plt.bar(["Costo Stock", "Riesgo Paro"], [costo_stock, riesgo_paro], color=["#1D3557","#E63946"]); plt.show()
```
