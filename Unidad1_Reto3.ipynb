```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 3
descuento_precio_pct = 10 #@param {type:"slider", min:0, max:25, step:1}
inversion_experiencia_pct = 15 #@param {type:"slider", min:0, max:30, step:1}

fidelidad = (descuento_precio_pct * 1.5) + (inversion_experiencia_pct * 2.2)
print(f"Índice de Fidelización Estimado: {min(100, fidelidad):.1f} / 100")
plt.bar(["Fidelidad Proyectada"], [min(100, fidelidad)], color="#1D3557")
plt.ylim(0, 100); plt.show()
```
