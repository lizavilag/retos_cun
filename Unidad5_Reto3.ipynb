```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 3
rebalanceo_pct = 35 #@param {type:"slider", min:0, max:70, step:5}

ingreso_infantil = 50000 * (1 - rebalanceo_pct/100) * 0.9
ingreso_mascotas = 50000 * (rebalanceo_pct/100) * 1.4

print(f"Ingreso Total Rebalanceado: ${ingreso_infantil + ingreso_mascotas:.2f}")
plt.bar(["Línea Infantil", "Línea Mascotas"], [ingreso_infantil, ingreso_mascotas], color=["#1D3557","#00A859"]); plt.show()
```
