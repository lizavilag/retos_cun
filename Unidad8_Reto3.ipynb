```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control Interactivo
parametro_1 = 50 #@param {type:"slider", min:0, max:100, step:5}
parametro_2 = 100 #@param {type:"number"}

resultado = (parametro_1 / 100) * parametro_2
print(f'Resultado Proyectado: {resultado:.2f}')

plt.bar(['Indicador Base', 'Resultado'], [parametro_2, resultado], color=['#1D3557', '#00A859'])
plt.ylabel('Valor')
plt.show()
```
