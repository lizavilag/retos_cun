```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 3
bono_usd = 60 #@param {type:"slider", min:10, max:150, step:10}

tasa_retoma = min(85, (bono_usd / 120) * 100)
equipos = int(1000 * (tasa_retoma/100))
print(f"Tasa de Retoma: {tasa_retoma:.1f}% | Equipos Reacondicionados: {equipos}")
plt.pie([tasa_retoma, 100-tasa_retoma], labels=["Recuperados", "No Devueltos"], colors=["#00A859","#E63946"], autopct="%1.1f%%"); plt.show()
```
