```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 2
precio_intermediario = 12 #@param {type:"number"}
precio_acopio = 22 #@param {type:"number"}
toneladas = 50 #@param {type:"slider", min:5, max:200, step:5}

ganancia_trad = toneladas * precio_intermediario
ganancia_acopio = toneladas * precio_acopio

print(f"Ganancia Tradicional: ${ganancia_trad} | Ganancia con Acopio: ${ganancia_acopio}")
plt.bar(["Intermediario", "Centro Acopio"], [ganancia_trad, ganancia_acopio], color=["#E63946", "#00A859"]); plt.show()
```
