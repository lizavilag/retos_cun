```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 1
produccion_unidades = 10000 #@param {type:"number"}
reconversion_bio_pct = 40 #@param {type:"slider", min:0, max:100, step:10}

impuesto = (produccion_unidades * (1 - reconversion_bio_pct/100)) * 0.15 * 10
print(f"Costo por Impuesto Plástico: ${impuesto:.2f}")
plt.pie([reconversion_bio_pct, 100-reconversion_bio_pct], labels=["Biodegradable", "Plástico Gravado"], colors=["#00A859","#E63946"], autopct="%1.1f%%"); plt.show()
```
