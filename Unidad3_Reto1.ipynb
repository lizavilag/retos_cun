```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 1
comision_aliado_pct = 15 #@param {type:"slider", min:5, max:30, step:1}
volumen_servicios = 300 #@param {type:"number"}

ingreso_red = volumen_servicios * 80 * (comision_aliado_pct/100)
print(f"Ingreso Adicional Red: ${ingreso_red:.2f}")
plt.bar(["Ingreso Red Aliados"], [ingreso_red], color="#00A859"); plt.show()
```
