```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 2
inflacion_pct = 10 #@param {type:"slider", min:2, max:25, step:1}
recorte_gastos_pct = 8 #@param {type:"slider", min:0, max:30, step:1}

ingresos = 100000 * (1 - inflacion_pct/100)
gastos = 70000 * (1 - recorte_gastos_pct/100)
flujo_neto = ingresos - gastos

print(f"Flujo Neto Proyectado: ${flujo_neto:.2f}")
plt.bar(["Ingresos", "Gastos", "Flujo Neto"], [ingresos, gastos, flujo_neto], color=["#1D3557","#E63946","#00A859"]); plt.show()
```
