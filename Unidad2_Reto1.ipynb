```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 1
capacidad_ventas = 200 #@param {type:"number"}
capacidad_despacho = 80 #@param {type:"slider", min:50, max:250, step:10}

pedidos_retrasados = max(0, capacidad_ventas - capacidad_despacho)
eficiencia = (min(capacidad_ventas, capacidad_despacho) / capacidad_ventas) * 100

print(f"Pedidos Represados: {pedidos_retrasados} | Eficiencia del Sistema: {eficiencia:.1f}%")
plt.bar(["Ventas", "Despacho Real"], [capacidad_ventas, capacidad_despacho], color=["#1D3557", "#E63946" if pedidos_retrasados>0 else "#00A859"])
plt.show()
```
