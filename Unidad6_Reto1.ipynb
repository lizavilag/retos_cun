```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 1
inversion_digital = 500 #@param {type:"slider", min:100, max:2000, step:100}

ventas_adic = inversion_digital * 3.8
roi = ((ventas_adic - inversion_digital) / inversion_digital) * 100
print(f"Ventas Adicionales: ${ventas_adic:.2f} | ROI: {roi:.1f}%")
plt.bar(["Inversión", "Ventas Generadas"], [inversion_digital, ventas_adic], color=["#1D3557","#00A859"]); plt.show()
```
