# 📊 SIMULADOR EN VIVO CUN: Análisis de Incremento de Costos por Proveedor
**Asignatura:** Entornos Organizacionales | **Unidad 1 - Reto 1**
---

```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 1
costo_base = 100 #@param {type:"number"}
aumento_prov_pct = 35 #@param {type:"slider", min:0, max:60, step:5}
ajuste_precio_pct = 15 #@param {type:"slider", min:0, max:40, step:1}

nuevo_costo = costo_base * (1 + aumento_prov_pct/100)
precio_venta = 150 * (1 + ajuste_precio_pct/100)
margen = ((precio_venta - nuevo_costo) / precio_venta) * 100

print(f"Nuevo Costo: ${nuevo_costo:.2f} | Nuevo Precio: ${precio_venta:.2f} | Margen Resultante: {margen:.1f}%")
plt.bar(["Margen Resultante (%)"], [margen], color="#00A859" if margen>=20 else "#E63946")
plt.axhline(20, color="gray", linestyle="--", label="Margen Mínimo (20%)")
plt.ylim(0, 50); plt.legend(); plt.show()
```
