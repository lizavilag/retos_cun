```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 2
horas_capacitacion_semana = 10 #@param {type:"slider", min:2, max:30, step:2}

semanas = 120 / horas_capacitacion_semana
print(f"Semanas estimadas para certificación ISO: {semanas:.1f} semanas")
plt.plot([0, semanas], [0, 100], marker="o", color="#00A859")
plt.xlabel("Semanas"); plt.ylabel("% Cumplimiento ISO"); plt.grid(True); plt.show()
```
