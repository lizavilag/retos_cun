```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 3
monto_proyecto = 50000 #@param {type:"number"}
cofinanciacion_pct = 60 #@param {type:"slider", min:20, max:80, step:5}

aporte_pyme = monto_proyecto * (1 - cofinanciacion_pct/100)
aporte_gob = monto_proyecto * (cofinanciacion_pct/100)

print(f"Aporte Pyme: ${aporte_pyme} | Fondo Público: ${aporte_gob}")
plt.pie([aporte_pyme, aporte_gob], labels=["Inversión Pyme", "Subsidio Público"], colors=["#1D3557","#00A859"], autopct="%1.1f%%"); plt.show()
```
