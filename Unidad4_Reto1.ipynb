```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 1
actores_integrados = 12 #@param {type:"slider", min:2, max:30, step:1}
gasto_diario = 80 #@param {type:"number"}

dias_estancia = 1.5 + (actores_integrados * 0.15)
derrama = dias_estancia * gasto_diario

print(f"Estancia Promedio: {dias_estancia:.1f} días | Derrama por Turista: ${derrama:.2f}")
plt.bar(["Derrama por Turista"], [derrama], color="#00A859"); plt.show()
```
