```python
import matplotlib.pyplot as plt

# @title 🎛️ Panel de Control del Simulador: Reto 1
dias_asia = 45 #@param {type:"number"}
dias_nearshoring = 8 #@param {type:"slider", min:3, max:15, step:1}

ahorro_dias = dias_asia - dias_nearshoring
print(f"Días Ahorrados: {ahorro_dias} días")
plt.bar(["Asia", "Nearshoring"], [dias_asia, dias_nearshoring], color=["#E63946","#00A859"]); plt.ylabel("Días de Tránsito"); plt.show()
```
